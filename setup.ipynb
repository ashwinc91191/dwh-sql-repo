```python
import subprocess
import sys
from pathlib import Path

# Set your Databricks project directory
BASE = Path("/Volumes/workspace/default/projectsetup")

# 1. Create all required folders
folders = [
    "script",
    "landing",
    "data/processing",
    "data/archive",
    "data/error",
    "data/output"
]

for folder in folders:
    (BASE / folder).mkdir(parents=True, exist_ok=True)

# 2. Create requirements.txt
req_file = BASE / "requirements.txt"

if not req_file.exists():
    req_file.write_text(
        "pandas\n"
        "openpyxl\n"
        "python-dotenv\n"
        "google-genai\n"
    )

# 3. Create .env placeholder
env_file = BASE / ".env"

if not env_file.exists():
    env_file.write_text(
        "GEMINI_API_KEY=PASTE_YOUR_GEMINI_API_KEY_HERE\n"
    )

# 4. Create initial script placeholder
script_file = BASE / "script" / "data_profiler.py"

if not script_file.exists():
    script_file.write_text(
        'import sys\n'
        'print("Running profile on:", '
        'sys.argv[1] if len(sys.argv) > 1 else "No file")\n'
    )

# 5. Install packages
print("Installing requirements...")

subprocess.run(
    [sys.executable, "-m", "pip", "install", "-r", str(req_file)],
    check=True
)

print("\nSetup completed successfully!")
print(f"Project location: {BASE}")
```
